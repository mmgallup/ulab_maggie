**Orbital Periods and Kepler's Third Law** \
Kepler's Third Law states that the square of the orbital period $T$ (in years) of a planet is proportional to the cube of the semi-major axis $a$ (in astronomical units, AU):

$$
T^2 = a^3
$$

Write a **function** that calculates the orbital period $T$ given the semi-major axis $a$. (So, you can assume that $T = \sqrt{a^3}$.)

```python
def orbital_period(a):
    # Calculate orbital period with Kepler's third law
    period = (a **3) ** 0.5
    return period
```

## 2 Indexing + Conditionals

You are provided with a list of exoplanet radii (in Earth radii) and their equilibrium temperatures (in Kelvin):

`radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]` \
`temperatures = [300, 400, 350, 280, 500, 450]`

Write a loop to filter and print the radii of planets with temperatures below 350 K.

```python
# Indexing and conditionals
radii = [1.2, 0.8, 1.5, 2, 1, 0.6]
temperatures = [300, 400, 350, 280, 500, 450] # Kelvin

# Find the planets with temperatures below 350 K
def find_planets(temperatures, radii):
    filtered_radii = []
    for i in range(len(temperatures)):
        if temperatures[i] < 350:
            filtered_radii.append(radii[i])
    return filtered_radii

find_planets(temperatures, radii)
```

## 4 Plotting
Create a scatter plot of the following semi-major axes (x-axis) vs. orbital periods (y-axis). Label the axes, add a legend and add a title: "Orbital Periods of Exoplanets."

`semi_major_axes = [0.5, 1.0, 2.0, 1.5, 1.2, 0.8]`

```python
import matplotlib.pyplot as plot

# semi-major axis values
semi_major_axis = [0.5, 1.0, 2.0, 1.5, 1.2, 0.8]

# compute the orbital periods
orbital_periods = np.sqrt(np.array(semi_major_axis) ** 3)

plot.scatter(semi_major_axis, orbital_periods, label='Orbital Period')

# add neccessary features to plot
plot.xlabel('Semi-Major Axis')
plot.ylabel('Orbital Period')
plot.title('Orbital Periods of Exoplanets')
plot.legend()

# create the plot
plot.show()
```

Create a bar plot of planet radii vs. their equilibrium temperatures. Label the axes, add a legend and add a title: "Equilibrium Temperatures vs. Radii."

```python
# Planet radii + corresponding equilibrium temperatures
radii = [1.2, 0.8, 1.5, 2, 1, 0.6]
temperatures = [300, 400, 350, 280, 500, 450]

plot.bar(radii, temperatures, label='Equilibrium Temperature', color='purple')

# add neccessary titles
plot.xlabel('Planet Radius (In Earth Radii)')
plot.ylabel('Equilibrium Temperature')
plot.title('Equilibrium Temperatures vs. Radii')
plot.legend()

# show plot
plot.show()
```

Put the scatter plot and bar plot on the same figure (horizonatally). 

```python
fig, ax = plot.subplots(1, 2, figsize=(12, 6))

# Scatter plot on left side
ax[0].scatter(semi_major_axes, orbital_periods, label='Orbital Period', color='purple')
ax[0].set_xlabel('Semi-Major Axis')
ax[0].set_ylabel('Orbital Period')
ax[0].set_title('Orbital Periods of Exoplanets')
ax[0].legend()

# Bar plot right side
ax[1].bar(radii, temperatures, label='Equilibrium Temperature', color='skyblue')
ax[1].set_xlabel('Planet Radius (In Earth Radii)')
ax[1].set_ylabel('Equilibrium Temperature')
ax[1].set_title('Equilibrium Temperatures vs. Radii')
ax[1].legend()

# Show plots
plot.tight_layout()
plot.show()
```
