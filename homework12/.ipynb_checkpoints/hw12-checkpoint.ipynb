## 2 Plotting
Make a scatter plot of **exoplanet Earth radius vs. equilibrium temperatures**. Include units, x/y axis labels, and a legend.

```python
import matplotlib.pyplot as plt

df = pd.read_csv('exoplanet_data.csv', comment='#')
planet_data = df[['pl_name', 'pl_rade', 'pl_eqt']]

planet_data = planet_data.dropna(subset=['pl_rade', 'pl_eqt'])

# Create a scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(planet_data['pl_rade'], planet_data['pl_eqt'], color='blue', label='Exoplanets')

# Add labels and title and grid etc...
plt.xlabel('Earth Radius', fontsize=12)
plt.ylabel('Equilibrium Temperature', fontsize=12)
plt.title('Exoplanet Earth Radius v Equilibrium Temperature', fontsize=14)
plt.grid(True)
plt.legend(loc='upper right')

# Show the plot
plt.show()
```

What trend do you notice in the data? Does it have positive or negative correlation?

```python
# It has a positive correlation. As x grows, so does y.
```

In the paper, find the age of the planet. What is the age, with error, of the planet?

```python
age_info = {
    'age': 3.5,  # in Gyrs
    'age_error_plus': 0.8,  # positive margin of error
    'age_error_minus': 1.3  # negative margin of error
}

print(f"Age: {age_info['age']} Gyr")
print(f"Margin of error: +{age_info['age_error_plus']} Gyr, -{age_info['age_error_minus']} Gyrs")
```

Create a new column called `Age` with your new variable and assign it the proper value. 

```python
df = pd.read_csv('exoplanet_data.csv')

# Define the Age + margin of error
age = 3.5
margin_plus = 0.8
margin_minus = 1.3 

# Create the new Age column
df['Age'] = age

df['Age_error_plus'] = margin_plus
df['Age_error_minus'] = margin_minus

# Save the updated DataFrame
df.to_csv('exoplanet_data_updated.csv', index=False)

print("New column 'Age' has been added.")
```

## 4 Make a New `.csv` File

To a new `.csv` file, save the planet WITHOUT the Nan rows and WITH the `Age` column added.
