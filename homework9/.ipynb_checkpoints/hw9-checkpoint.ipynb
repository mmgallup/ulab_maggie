# 2 Plotting a Function
Goal: Write a **function** that will plot $f(x) = x^2$ and $g(x) = \sqrt{x}$ on the same plot and return the $f(x)$ and $g(x)$ values as separate lists. Your plot needs to have the following:
- An appropriate figure size (`figsize=(10,8)` is recommended)
- Labeled x and y axis
- A title
- Different colors and linestyles for each function
- A legend
- Your domain should be $x \in [0, 10]$. *Hint: Use `np.linspace` to set a range of values for your functions.*

Your function should return:
- A list of $f(x)$ values
- A list of $g(x)$ values.

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_functions():
    # Plots f(x) = x^2 and g(x) = sqrt(x) on the same plot and returns the f(x) and g(x) values as separate lists.
    # Returns: f_values (list) and g_values (list)
    
    x_values = np.linspace(0, 10, 100)
    
    f_values = x_values ** 2
    g_values = np.sqrt(x_values)
    
    plt.figure(figsize=(10, 8))
    plt.plot(x_values, f_values, color="blue", linestyle="-", label="f(x) = x^2")
    plt.plot(x_values, g_values, color="green", linestyle="--", label="g(x) = sqrt(x)")
    plt.xlabel("x-axis")
    plt.ylabel("y-axis")
    plt.title('Plot of f(x) = x^2 and g(x) = sqrt(x)')
    plt.legend()
    plt.show()
    
    return f_values.tolist(), g_values.tolist()

f_vals, g_vals = plot_functions() 
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_functions_with_all_features():
    # Plots f(x) = x^2 and g(x) = sqrt(x) on the same plot with customizations including:
    # - Vertical and horizontal lines
    # - Custom axis limits
    # - Change domain of x
    # - Legend outside the plot
    # - Returns the f(x), g(x), and their sum values
    
    x_values = np.linspace(0, 10, 100)
    
    f_values = x_values ** 2
    g_values = np.sqrt(x_values)
    
    sum_values = f_values + g_values
    
    plt.figure(figsize=(10, 8))
    plt.plot(x_values, f_values, color='blue', linestyle='-', label='f(x) = x^2')
    plt.plot(x_values, g_values, color='green', linestyle='--', label='g(x) = sqrt(x)')
    plt.axvline(x=5, color='red', linestyle='-.', label='Vertical Line at x=5')
    plt.axhline(y=25, color='purple', linestyle=':', label='Horizontal Line at y=25')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('Plot of f(x) = x^2 and g(x) = sqrt(x) with Custom Features')
    plt.xlim(0, 8)  # Limits x-axis from 0 to 8
    plt.ylim(0, 70)  # Limits y-axis from 0 to 70
    plt.legend(loc='upper right', bbox_to_anchor=(1.15, 1))
    plt.show()
    
    return f_values.tolist(), g_values.tolist(), sum_values.tolist()

f_vals, g_vals, sum_vals = plot_functions_with_all_features()
```

Change the domain of your function to be **twice** a large as the previous domain. Call both of your functions in the cell below with the new domain.

```python
import function_plots

def plot_horizontal_subplots_with_custom_domain(x_values):
    # Adds a new domain to the old module
    
    h_values = np.cos(x_values)
    k_values = np.sin(x_values)
    plt.figure(figsize=(14, 6))
    plt.subplot(1, 2, 1)
    plt.plot(x_values, h_values, color="blue", linestyle="-", label="h(x) = cos(x)")
    plt.xlabel("x")
    plt.ylabel("h(x)")
    plt.title("Plot of h(x) = cos(x)")
    plt.legend()
    plt.subplot(1, 2, 2)
    plt.plot(x_values, k_values, color="green", linestyle="--", label="k(x) = sin(x)")
    plt.xlabel("x")
    plt.ylabel("k(x)")
    plt.title("Plot of k(x) = sin(x)")
    plt.legend()
    plt.tight_layout()
    plt.show()

def plot_vertical_subplots_with_custom_domain(x_values):
    # Adds another new domain to the old module

    h_values = np.cos(x_values)
    k_values = np.sin(x_values)
    plt.figure(figsize=(10, 12))
    plt.subplot(2, 1, 1)
    plt.plot(x_values, h_values, color='blue', linestyle='-', label='h(x) = cos(x)')
    plt.xlabel('x')
    plt.ylabel('h(x)')
    plt.title('Plot of h(x) = cos(x)')
    plt.legend()
    plt.subplot(2, 1, 2)
    plt.plot(x_values, k_values, color='green', linestyle='--', label='k(x) = sin(x)')
    plt.xlabel('x')
    plt.ylabel('k(x)')
    plt.title('Plot of k(x) = sin(x)')
    plt.legend()
    plt.tight_layout()
    plt.show()

# Defining the new domain
new_x_values = np.linspace(0, 4 * np.pi, 100)

plot_horizontal_subplots_with_custom_domain(new_x_values)
plot_vertical_subplots_with_custom_domain(new_x_values)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def plot_3d_sphere():
    # This function plots a 3D sphere, yay!

    u = np.linspace(0, 2 * np.pi, 100)
    v = np.linspace(0, np.pi, 100)
    u, v = np.meshgrid(u, v)

    x = np.sin(v) * np.cos(u)
    y = np.sin(v) * np.sin(u)
    z = np.cos(v)

    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    ax.plot_surface(x, y, z, cmap='viridis')

    ax.set_xlabel("X-axis")
    ax.set_ylabel("Y-axis")
    ax.set_zlabel("Z-axis")
    ax.set_title("3D Sphere Plot!")
    plt.legend()

    plt.show()

plot_3d_sphere()
```
